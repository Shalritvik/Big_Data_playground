Predicting EV Charger Availability in NYC
End-to-end Spark pipeline + EDA + ML + ROC

```python
# Imports + Spark setup

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.functions import col, when, to_date, year, current_date, dayofweek, regexp_replace, trim, expr

from pyspark.ml import Pipeline
from pyspark.ml.feature import StringIndexer, OneHotEncoder, VectorAssembler
from pyspark.ml.classification import LogisticRegression, RandomForestClassifier
from pyspark.ml.evaluation import BinaryClassificationEvaluator, MulticlassClassificationEvaluator
from pyspark.ml.functions import vector_to_array
```

```python
sns.set_theme(style="whitegrid")

spark = SparkSession.builder.appName("EV_Charging_Availability_NYC").getOrCreate()
spark.conf.set("spark.sql.ansi.enabled", "false")
```

 Load data

```python
file_path = "/content/Electric_Vehicle_Charging_Stations_in_New_York_20251120.csv"
sdf = spark.read.csv(file_path, header=True, inferSchema=True)

print("Initial row count:", sdf.count())
sdf.printSchema()
```

 3) Drop irrelevant columns

```python
columns_to_drop = [
    "Expected Date",
    "Federal Agency ID",
    "Federal Agency Name",
    "Fuel Type Code"   # we will filter to electric via connector/network fields below instead
]
existing_cols_to_drop = [c for c in columns_to_drop if c in sdf.columns]
sdf = sdf.drop(*existing_cols_to_drop)

# Remove duplicates
sdf = sdf.dropDuplicates()

print("After dropDuplicates row count:", sdf.count())

```

4) Defensive casting + cleaning

```python
def safe_double(colname: str):  # safe_double(colname) returns a Spark SQL expression that tries to convert a column into a double safely
    return expr(f"try_cast(nullif(trim(`{colname}`), '') as double)")

# Cast key numeric columns
for c in ["Latitude", "Longitude", "EV Level1 EVSE Num", "EV Level2 EVSE Num", "EV DC Fast Count"]: #For each listed column, if it exists, it overwrites that column with the safely casted double version.
    if c in sdf.columns:
        sdf = sdf.withColumn(c, safe_double(c))

# ZIP: digits only -> int
if "ZIP" in sdf.columns:
    sdf = sdf.withColumn(
        "ZIP",
        expr("try_cast(nullif(regexp_replace(cast(`ZIP` as string), '[^0-9]', ''), '') as int)")
    )

# Parse dates safely
if "Open Date" in sdf.columns:
    sdf = sdf.withColumn("Open Date", to_date(col("Open Date")))
if "Date Last Confirmed" in sdf.columns:
    sdf = sdf.withColumn("Date Last Confirmed", to_date(col("Date Last Confirmed")))

# Fill numeric nulls
fill_numeric = {
    "EV Level1 EVSE Num": 0.0,
    "EV Level2 EVSE Num": 0.0,
    "EV DC Fast Count": 0.0,
    "Latitude": 0.0,
    "Longitude": 0.0
}
sdf = sdf.fillna(fill_numeric)

# Keep only "open" stations if Status Code exists (common in this dataset)
if "Status Code" in sdf.columns:
    sdf = sdf.filter((col("Status Code") == "E") | col("Status Code").isNull())

# Drop rows missing location essentials (avoid empty plots / broken features)
sdf = sdf.filter(col("Latitude").isNotNull() & col("Longitude").isNotNull())

print("Cleaned row count:", sdf.count())
sdf.printSchema()
```

5) Feature engineering for "availability"

```python
# This dataset does not include real-time occupancy ("busy/available").
# So we define an infrastructure-based availability proxy consistent with your report narrative:
# - higher capacity (more ports, especially DC fast) => more likely "available" / less congestion risk.
# - add access constraints as a proxy for availability.

# Station age (temporal infrastructure trend)
sdf = sdf.withColumn("Open Year", year(col("Open Date")))
sdf = sdf.withColumn("station_age", when(col("Open Year").isNotNull(),
                                        year(current_date()) - col("Open Year")).otherwise(None))

# Access proxy (24/7 vs restricted)
if "Access Days Time" in sdf.columns:
    sdf = sdf.withColumn(
        "access_type",
        when(col("Access Days Time").rlike("24"), "24/7").otherwise("Restricted")
    )
else:
    sdf = sdf.withColumn("access_type", F.lit("Unknown"))

# Capacity score (DC fast weighted higher)
sdf = sdf.withColumn(
    "capacity_score",
    (col("EV Level2 EVSE Num") + 2 * col("EV DC Fast Count")).cast("double")
)

# Robust quantile threshold (avoid IndexError)
quantiles = sdf.filter(col("capacity_score").isNotNull()).approxQuantile("capacity_score", [0.75], 0.01)
if len(quantiles) == 0:
    # Fallback threshold if data is weird: use mean
    threshold = sdf.select(F.avg("capacity_score")).first()[0]
else:
    threshold = quantiles[0]

print("Availability threshold (capacity_score 75th pct or fallback):", threshold)

# Label: high availability (1) vs low availability (0)
sdf = sdf.withColumn("label", when(col("capacity_score") >= F.lit(threshold), 1).otherwise(0))

# Ensure label is non-null
sdf = sdf.na.drop(subset=["label"])

sdf.groupBy("label").count().show()

```

6) EDA Visualizations

```python
# Convert to pandas only for small aggregations (safe)
# Figure 1: Top 10 Cities
if "City" in sdf.columns:
    top_cities_pd = (
        sdf.groupBy("City").count().orderBy(col("count").desc()).limit(10).toPandas()
    )
    plt.figure(figsize=(12, 6))
    sns.barplot(data=top_cities_pd, x="count", y="City")
    plt.title("Top 10 Cities by Number of Charging Stations")
    plt.xlabel("Number of Stations")
    plt.ylabel("City")
    plt.show()

# Figure 2: Top 10 EV Networks
if "EV Network" in sdf.columns:
    top_networks_pd = (
        sdf.groupBy("EV Network").count().orderBy(col("count").desc()).limit(10).toPandas()
    )
    plt.figure(figsize=(12, 6))
    sns.barplot(data=top_networks_pd, x="count", y="EV Network")
    plt.title("Top 10 EV Charging Networks")
    plt.xlabel("Number of Stations")
    plt.ylabel("Network")
    plt.show()

# Figure 3: Connector Types Distribution (Top 10)
if "EV Connector Types" in sdf.columns:
    top_connectors_pd = (
        sdf.groupBy("EV Connector Types").count().orderBy(col("count").desc()).limit(10).toPandas()
    )
    plt.figure(figsize=(12, 6))
    sns.barplot(data=top_connectors_pd, x="count", y="EV Connector Types")
    plt.title("Top 10 EV Connector Configurations")
    plt.xlabel("Count")
    plt.ylabel("Connector Type")
    plt.show()

# Extra (recommended): 24/7 vs Restricted
access_pd = sdf.groupBy("access_type").count().toPandas()
plt.figure(figsize=(7, 4))
sns.barplot(data=access_pd, x="access_type", y="count")
plt.title("Station Access Type (Availability Proxy)")
plt.xlabel("Access Type")
plt.ylabel("Number of Stations")
plt.show()
```

7) ML Pipeline features

```python
categorical_cols = [c for c in ["City", "EV Network", "EV Connector Types", "access_type"] if c in sdf.columns]

numeric_cols = []
for c in ["Latitude", "Longitude", "EV Level2 EVSE Num", "EV DC Fast Count", "station_age", "capacity_score"]:
    if c in sdf.columns:
        numeric_cols.append(c)

# Capacity_score is derived from EV Level2 + EV DC Fast (still OK, since label is threshold on it).
# Fill any remaining numeric nulls
sdf = sdf.fillna(0.0, subset=numeric_cols)

indexers = [
    StringIndexer(inputCol=c, outputCol=f"{c}_idx", handleInvalid="keep")
    for c in categorical_cols
]

encoder = OneHotEncoder(
    inputCols=[f"{c}_idx" for c in categorical_cols],
    outputCols=[f"{c}_ohe" for c in categorical_cols]
)

assembler = VectorAssembler(
    inputCols=numeric_cols + [f"{c}_ohe" for c in categorical_cols],
    outputCol="features",
    handleInvalid="keep"
)

# Models
lr = LogisticRegression(
    featuresCol="features",
    labelCol="label",
    maxIter=50
)

rf = RandomForestClassifier(
    featuresCol="features",
    labelCol="label",
    numTrees=200,
    maxDepth=8,
    seed=42
)

lr_pipeline = Pipeline(stages=indexers + [encoder, assembler, lr])
rf_pipeline = Pipeline(stages=indexers + [encoder, assembler, rf])
```

8) Train/test split

```python
train_df, test_df = sdf.randomSplit([0.8, 0.2], seed=42)
print("Train count:", train_df.count())
print("Test count:", test_df.count())

```

9) Evaluation helper

```python
def train_and_eval(pipeline, name):
    print(f"\n=== {name} ===")
    model = pipeline.fit(train_df)
    preds = model.transform(test_df)

    # Accuracy
    acc_eval = MulticlassClassificationEvaluator(
        labelCol="label",
        predictionCol="prediction",
        metricName="accuracy"
    )
    acc = acc_eval.evaluate(preds)
    err = 1 - acc

    # AUC
    auc_eval = BinaryClassificationEvaluator(
        labelCol="label",
        rawPredictionCol="rawPrediction",
        metricName="areaUnderROC"
    )
    auc = auc_eval.evaluate(preds)

    # Precision / Recall / F1
    prec_eval = MulticlassClassificationEvaluator(
        labelCol="label", predictionCol="prediction", metricName="weightedPrecision"
    )
    rec_eval = MulticlassClassificationEvaluator(
        labelCol="label", predictionCol="prediction", metricName="weightedRecall"
    )
    f1_eval = MulticlassClassificationEvaluator(
        labelCol="label", predictionCol="prediction", metricName="f1"
    )

    precision = prec_eval.evaluate(preds)
    recall = rec_eval.evaluate(preds)
    f1 = f1_eval.evaluate(preds)

    print(f"Accuracy:   {acc:.4f}")
    print(f"Error rate: {err:.4f}")
    print(f"Precision:  {precision:.4f}")
    print(f"Recall:     {recall:.4f}")
    print(f"F1:         {f1:.4f}")
    print(f"AUC:        {auc:.4f}")

    # Confusion matrix
    tp = preds.filter((col("label") == 1) & (col("prediction") == 1)).count()
    tn = preds.filter((col("label") == 0) & (col("prediction") == 0)).count()
    fp = preds.filter((col("label") == 0) & (col("prediction") == 1)).count()
    fn = preds.filter((col("label") == 1) & (col("prediction") == 0)).count()
    print(f"TP={tp}, TN={tn}, FP={fp}, FN={fn}")

    return model, preds, auc, acc

lr_model, lr_preds, lr_auc, lr_acc = train_and_eval(lr_pipeline, "Logistic Regression")
rf_model, rf_preds, rf_auc, rf_acc = train_and_eval(rf_pipeline, "Random Forest")

```

```python
rf_scores = rf_preds.select(
    col("label").alias("label"),
    vector_to_array("probability")[1].alias("score")
).toPandas()

from sklearn.metrics import roc_curve, roc_auc_score

fpr, tpr, _ = roc_curve(rf_scores["label"], rf_scores["score"])
auc_val = roc_auc_score(rf_scores["label"], rf_scores["score"])

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label=f"Random Forest (AUC={auc_val:.3f})")
plt.plot([0, 1], [0, 1], "k--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve – Charger Availability (Infrastructure Proxy)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta


#  Loading the dummy dataset

stations = pd.read_csv(
    "Electric_Vehicle_Charging_Stations_in_New_York_20251120.csv"
)

# Keep only essential columns
stations = stations[[
    "ID", "Station Name", "City", "ZIP",
    "Latitude", "Longitude",
    "EV Level2 EVSE Num", "EV DC Fast Count"
]].dropna(subset=["Latitude", "Longitude"])

# Sample stations to keep dataset reasonable
stations = stations.sample(frac=0.4, random_state=42)

# Generate synthetic sessions

sessions = []

for _, row in stations.iterrows():
    num_sessions = random.randint(20, 60)

    for _ in range(num_sessions):
        # Day type
        day_type = random.choices(
            ["Weekday", "Weekend"],
            weights=[0.7, 0.3]
        )[0]

        # Hour distribution
        if day_type == "Weekday":
            hour = random.choices(
                list(range(24)),
                weights=[1]*7 + [3]*3 + [1]*6 + [4]*4 + [1]*4
            )[0]
        else:
            hour = random.choices(
                list(range(24)),
                weights=[1]*9 + [3]*8 + [1]*7
            )[0]

        charger_type = (
            "DC Fast" if row["EV DC Fast Count"] > 0 and random.random() < 0.4
            else "Level 2"
        )

        duration = (
            random.randint(15, 45) if charger_type == "DC Fast"
            else random.randint(60, 180)
        )

        energy = (
            round(random.uniform(20, 60), 1) if charger_type == "DC Fast"
            else round(random.uniform(5, 20), 1)
        )

        start_time = datetime(2024, random.randint(1, 12), random.randint(1, 28), hour)

        sessions.append({
            "station_id": row["ID"],
            "station_name": row["Station Name"],
            "city": row["City"],
            "zip": row["ZIP"],
            "latitude": row["Latitude"],
            "longitude": row["Longitude"],
            "charger_type": charger_type,
            "session_start": start_time,
            "hour_of_day": hour,
            "day_type": day_type,
            "session_duration_min": duration,
            "energy_kwh": energy,
            "is_peak_hour": hour in [7,8,9,16,17,18,19]
        })


# Create & save CSV

synthetic_df = pd.DataFrame(sessions)
synthetic_df.to_csv(
    "synthetic_ev_charging_sessions_nyc.csv",
    index=False
)

print("Synthetic dataset created:")
print(synthetic_df.head())
print("Total sessions:", len(synthetic_df))
```

Haversine function : The haversine formula determines the great-circle distance between two points on a sphere given their longitudes and latitudes.

```python
import math

def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Earth radius in km
    phi1, phi2 = math.radians(lat1), math.radians(lat2)
    dphi = math.radians(lat2 - lat1)
    dlambda = math.radians(lon2 - lon1)

    a = math.sin(dphi/2)**2 + math.cos(phi1)*math.cos(phi2)*math.sin(dlambda/2)**2
    return 2 * R * math.atan2(math.sqrt(a), math.sqrt(1 - a))
```

```python
stations_pdf = sdf.select(
    "Station Name",
    "City",
    "ZIP",
    "Latitude",
    "Longitude",
    "EV Level2 EVSE Num",
    "EV DC Fast Count",
    "EV Connector Types",
    "EV Network"
).dropna(subset=["Latitude", "Longitude", "ZIP"]).toPandas()
```

```python
zip_centers = (
    stations_pdf.groupby("ZIP")[["Latitude", "Longitude"]]
    .mean()
    .to_dict("index")
)
```

```python
def find_nearest_station_by_zip(user_zip, k=3):
    user_zip = int(user_zip)

    if user_zip not in zip_centers:
        print("No stations found for that ZIP code.")
        return None

    user_lat = zip_centers[user_zip]["Latitude"]
    user_lon = zip_centers[user_zip]["Longitude"]

    stations_pdf["distance_km"] = stations_pdf.apply(
        lambda row: haversine(
            user_lat, user_lon, row["Latitude"], row["Longitude"]
        ),
        axis=1
    )

    return stations_pdf.sort_values("distance_km").head(k)
```
